```python
PASS_MARK = 40
MAX_MARKS = 100

def get_grade(average):
    if average >= 90:
        return "A+"
    elif average >= 80:
        return "A"
    elif average >= 70:
        return "B"
    elif average >= 60:
        return "C"
    elif average >= 50:
        return "D"
    elif average >= PASS_MARK:
        return "E"
    return "F"


print(get_grade(85))
print(get_grade(45))
print(get_grade(30))
```

```python
def get_valid_marks(prompt):
    while True:
        try:
            marks = float(input(prompt))
            if 0 <= marks <= MAX_MARKS:
                return marks
            print(f"  Marks must be between 0 and {MAX_MARKS}.")
        except ValueError:
            print("  Please enter a valid number.")

m = get_valid_marks("Enter marks: ")
print("You entered:", m)
```

```python
def analyze_student(name, marks):
    total = sum(marks.values())
    average = total / len(marks)
    passed = all(m >= PASS_MARK for m in marks.values())
    return {
        "name": name,
        "marks": marks,
        "total": total,
        "average": average,
        "grade": get_grade(average) if passed else "F",
        "status": "Pass" if passed else "Fail",
    }

result = analyze_student("Anjali", {"Maths": 80, "Physics": 70, "English": 90})
print(result)

result2 = analyze_student("Test", {"Maths": 80, "Physics": 30, "English": 90})
print(result2["status"], result2["grade"])
```

```python
def collect_students():
    students = []

    subjects = input("Enter subjects separated by commas (e.g. Maths,Physics,English): ")
    subjects = [s.strip().title() for s in subjects.split(",") if s.strip()]
    if not subjects:
        subjects = ["Maths", "Science", "English"]
        print("No subjects entered, using:", ", ".join(subjects))

    count = int(get_valid_marks("How many students? "))

    for i in range(1, count + 1):
        print(f"\n--- Student {i} ---")
        name = input("Name: ").strip().title() or f"Student {i}"
        marks = {}
        for subject in subjects:
            marks[subject] = get_valid_marks(f"  {subject} marks: ")
        students.append(analyze_student(name, marks))

    return students, subjects

students, subjects = collect_students()
print(students)
```

```python
def print_report(students, subjects):
    line = "=" * (30 + 10 * len(subjects) + 32)
    print("\n" + line)
    print("STUDENT PERFORMANCE REPORT".center(len(line)))
    print(line)

    header = f"{'Name':<15}"
    for s in subjects:
        header += f"{s:>10}"
    header += f"{'Total':>9}{'Avg':>8}{'Grade':>7}{'Status':>8}"
    print(header)
    print("-" * len(line))

    for st in students:
        row = f"{st['name']:<15}"
        for s in subjects:
            row += f"{st['marks'][s]:>10.1f}"
        row += f"{st['total']:>9.1f}{st['average']:>8.2f}{st['grade']:>7}{st['status']:>8}"
        print(row)

    print(line)

    passed = [s for s in students if s["status"] == "Pass"]
    topper = max(students, key=lambda s: s["total"])
    class_avg = sum(s["average"] for s in students) / len(students)

    print(f"Total students : {len(students)}")
    print(f"Passed / Failed: {len(passed)} / {len(students) - len(passed)}")
    print(f"Class average  : {class_avg:.2f}")
    print(f"Topper         : {topper['name']} ({topper['total']:.1f} marks)")

    print("\nSubject-wise average:")
    for subject in subjects:
        avg = sum(s["marks"][subject] for s in students) / len(students)
        print(f"  {subject:<12}: {avg:.2f}")

print_report(students, subjects)
```
